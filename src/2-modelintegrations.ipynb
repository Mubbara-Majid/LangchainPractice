## Streaming

Streaming sends the model's output piece by piece (token by token) as soon as each part is ready, instead of waiting for the full response to finish generating.

```python
google_model.invoke("Write a 200 words essay on the uses of Artficial Intelligence in healthcare.")
```

```python
for chunk in google_model.stream("Write a 200 words essay on the uses of Artificial Intelligence in healthcare."):
    print(chunk, end="", flush=True)
```

```python
for chunk in google_model.stream("Why do parrots have colorful feathers?"):
    print(chunk, end="", flush=True)
```

## Batching

Batching groups multiple inputs together and processes them in parallel or sequentially in a single operation, rather than making separate calls one after another.

```python
prompts = [
    "Define gravity in one sentence.",
    "Define relativity in one sentence.",
    "Define entropy in one sentence."
]

responses = google_model.batch(prompts,
                                config={
                                    'max_concurrency': 5
                                })

for response in responses:
    print(response)
```
